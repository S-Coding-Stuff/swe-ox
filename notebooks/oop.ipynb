```python
import numpy as np

def attach_names(data, names):
    if len(data) != len(names):
        raise ValueError("Length of data and names must be the same.")
    return {name: row for name, row in zip(names, data)}

# def attach_names(data, names):
#     output = {}
#     for i, name in enumerate(names):
#         output[name] = data[i]
#     return output

data = np.array([[1.,2.,3.], 
                 [4.,5.,6.]])

output = attach_names(data, ['Alice', 'Bob'])
print(output)  # Output: {'Alice': array([1., 2., 3.]), 'Bob': array([4., 5., 6.])}
```

### Properties

```python
# file: inflammation/models.py

class Patient:
    """A patient in an inflammation study."""
    def __init__(self, name):
        self.name = name
        self.observations = []

    def add_observation(self, value, day=None):
        if day is None:
            try:
                day = self.observations[-1]['day'] + 1

            except IndexError:
                day = 0


        new_observation = {
            'day': day,
            'value': value,
        }

        self.observations.append(new_observation)
        return new_observation

    @property
    def last_observation(self):
        return self.observations[-1]

    def __str__(self):
        return self.name




alice = Patient('Alice')

alice.add_observation(3)
alice.add_observation(4)

obs = alice.last_observation
print(obs)
```

### Inheritance and Composition

```python
class Observation:
    def __init__(self, day, value):
        self.day = day
        self.value = value

    def __str__(self):
        return str(self.value)

class Patient:
    """A patient in an inflammation study."""
    def __init__(self, name):
        self.name = name
        self.observations = []

    def add_observation(self, value, day=None):
        if day is None:
            try:
                day = self.observations[-1].day + 1

            except IndexError:
                day = 0

        new_observation = Observation(day, value)

        self.observations.append(new_observation)
        return new_observation

    def __str__(self):
        return self.name


alice = Patient('Alice')
obs = alice.add_observation(3)

print(obs)
```

### Mixin

A mixin is a class that provides methods to other classes, but is not considered a base class itself. Essentially, the class is designed purely to be used in composition with other classes, rather than to be instantiated on its own. If you have repeated functionality that you want to share across multiple classes, you can define it in a mixin class and then include that mixin in the classes that need it. This is especially useful in languages, like Python or C++, that support multiple inheritance.

```python
class nameMixin:
    def __str__(self):
        return self.name
```

```python
class PatientWithMixin(nameMixin):
    def __init__(self, name):
        self.name = name
```

```python
# Adjustments made based on polymorphism

class Observation:
    def __init__(self, day, value):
        self.day = day
        self.value = value

    def __str__(self):
        return str(self.value)

class Person:
    def __init__(self, name):
        self.name = name
        self.id = None

    def __str__(self):
        return self.name

    def set_id(self, id):
        self.id = 'P' + str(id).zfill(4)

    def get_id(self, id):
        return self.id

class Patient(Person):
    """A patient in an inflammation study."""
    def __init__(self, name):
        super().__init__(name)
        self.observations = []
        self.id = None

    def add_observation(self, value, day=None):
        if day is None:
            try:
                day = self.observations[-1].day + 1

            except IndexError:
                day = 0

        new_observation = Observation(day, value)

        self.observations.append(new_observation)
        return new_observation

    def set_id(self, id):
        self.id = 'P' + str(id).zfill(4)

# Part of the practical
class Doctor(Person):
    def __init__(self, name):
        super.__init__(name)
        self.patients = []

    def add_patient(self, new_patient):
        for patient in self.patients:
            if patient.name == new_patient.name:
                return
        self.patients.append(new_patient)

    def set_id(self, id):
            self.id = 'D' + str(id).zfill(4)
```

```python
class Administrator:
    """An administrator in an inflammation study."""
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return 'Administrator: ' + self.name

    def set_id(self, id):
        self.id = 'A' + str(id).zfill(4)

    def get_id(self):
        return self.id
```

### Polymorphism

```python
class Trial:
    def __init__(self):
        self.people = []

    def add_person(self, person):
        if isinstance(person, Person):
            self.people.append(person)
        elif hasattr(person, 'get_id'):
            person_id = person.get_id()
            if person_id[0] == 'D':
                person = Doctor(person_id)
            elif person_id[0] == 'P':
                person = Patient(person_id)
            elif person_id[0] == 'A':
                person = Administrator(person_id)
            else:
                raise TypeError('Could not recognise person id, must start with D, P, or A!')
            self.people.append(person)
        else:
            raise TypeError("Person object must be relevant or have a get_id method!")

    def print_people(self):
        for person in self.people:
            print(person)

    def set_ids(self):
        for i , person in enumerate(self.people):
            person.set_id(i+1)
```
