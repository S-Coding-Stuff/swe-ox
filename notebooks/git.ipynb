# Version Control and Git

Made by the GOAT Linus Torvalds.

Records changes made to code files via pushing and pulling to an external server e.g. GitHub or *GitLab*. Essentially acts as a backup of your code.

Remember to *tag* key versions of your codebase. 

*Git* is most popular, *Jujutsu* promises a cleaner workflow and is Git compatible. 



### What **NOT** to put in a repo

- Generated output files
- Compiled binaries
- Packages/(virtual) environments
- IDE/OS Files
- API Keys
- `.env` files -> This can include personal data. Place any unwanted files in `.gitignore`

### Initialise Git Repository

```
git --version
git init
git clone <repository-url>

git config --global user.name "Your Name"
git config --global user.email "you@example.com"
git config --list
```

### Commands

- All Git history - `git log`; `git log --oneline --graph --decorate --all`;
- Line-by-line difference of two states - `git diff <commit-a> <commit-b>`
- Push - `git push <file>`
- Add and commit - `git add <file>`; `git commit -m "<message>"`; `git commit --amend`;
- Create a new branch - `git branch` or `git checkout -p`
- Merge - `git merge <branch>`; `git rebase <branch>`; `git rebase --continue`; `git rebase <abort>`
- Download remote updates without merge - `git fetch`
- Tagging - `git tage`; `git tag <tag-name>`; `git push origin <tag-name>`;
- Stash temporary work - `git stash`; `git stash list`; `git stash pop`; `git stash apply`;


### Undo Changes **(Important)**

``` 
git revert <commit>      # Create a new commit that reverses a commit
git reset --soft HEAD~1  # Undo latest commit, keep changes staged
git reset HEAD~1         # Undo latest commit, keep changes unstaged
git reflog               # Find recent HEAD positions and recover lost commits
```
