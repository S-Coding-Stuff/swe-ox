# Software Testing

Add tests to preserve functionality over time, while updating the codebase. This is something that should be started earlier since bugs become more expensive to fix later in the software cycle.

### Questions

- Does it work as expected?
- Can correctness be verified independently?
- How confident are we in the accuracy of the results?

### Automated Testing
Manual methods are time consuming and expensive, and not at all scalable when the code grows in size. Manual tests will still be required!

**Unit Tests:** Check specific unit of functionality for expected outputs from given inputs.

**Integration Tests:** Test functional paths between various pieces of code and interactions between components.

**Regression Tests:** Confirm that program outputs stay the same after modifications.

---

### Pytest
This is the standard testing framework in Python. Uses `assert` to check for expected results. Supports fixture for setup and teardown of test environments.

```python
@pytest.fixture
def sample_data():
    return [1,2,3]

def test_sum(sample_data):
    assert sum(sample_data) == 6
```

Allows parameterisation to run the same test with different inputs:

```python
@pytest.mark.parameterize("input", expected[
    ...
])
```

### Debugging
Unit tests can detect problems and narrow down where to look, but don't tell us what the source of the problem is.

Implement logging module in Python to aid in debugging rather than using `print()` statements. This traces execution with timestamps and configurable levels of detail.

Also need to examine any intermediate files. If all else fails use the debugger in VSCode or whatever IDE.

**Defensive Programming:** check that input data satisfies preconditions before continuing i.e. doing `isinstance()` or `istype()` and then raising an error given some conditions.

**Code Quality Checks:** Use *linters* to detect errors, inconsistencies, and style issues automatically e.g. `pylance` or `ruff`. These improve readability, maintainability, and consistency.

`Code Smells:` These signal deeper design issues even when code still runs e.g. oversized classes, many nested if statements (disgusting)m duplicate branches.
