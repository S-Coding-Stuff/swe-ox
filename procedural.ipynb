```python
# Class
class Academic:
    # Dunder method
    def __init__(self, name, papers=None):
        self.name = name
        self.papers = papers if papers is not None else []

    def write_papers(self, title):
        self.papers.append(title)

    # For private methods, we can use a single underscore prefix to indicate that they are intended for internal use only. However, this is just a convention and does not enforce any access restrictions.
```

```python
class PhotonSource:
    def generate(self):
        raise NotImplementedError("Subclasses must implement the generate method.")

class BlackHole(PhotonSource):
    def generate(self):
        return "Generating photons from a black hole."

class WhiteDwarf(PhotonSource):
    def generate(self):
        return "Generating photons from a white dwarf."
```
